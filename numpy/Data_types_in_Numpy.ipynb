#### Why Data Types Matter in Numpy

```python
arr_int64 = np.array([1, 2, 3], dtype = np.int64) 
arr_int32 = np.array([1, 2, 3], dtype = np.int32) 

print(arr_int64.nbytes) #output: 24 bytes (3 elemwnts * 8 bytes)
print(arr_int32.nbytes) #output: 12 bytes (3 elemwnts * 4 bytes)
```
