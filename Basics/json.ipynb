```python
import json
obj = {"name":"Alice", "company":"IBM", "salary":45000}
s = json.dumps(obj)
print(s)
print(type(s))
```

```python
d = json.loads('{"name": "Alice", "company": "IBM", "salary": 45000}')
print(d)
print(type(d))
```

```python
file = open("data.json", "w")
```

```python
json.dump(obj, file)
```
