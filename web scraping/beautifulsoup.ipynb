```python
total_books = 0
list_of_books = []
for i in range(1, 51):
    
    with open(f"htmls/page{1}.html", "r") as f:
        content = f.read()
        
    soup = BeautifulSoup(content, "html.parser")
    products = soup.select(".product_pod")
    
    for product in products[0:]:
        
        title = product.select('a')[1]['title']
        price = product.select_one('.price_color').text[4:]
        list_of_books.append([title, price])
        
        total_books += 1
print(f"Total books are : {total_books}")
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(list_of_books, columns = ["Book", "Price"])
```
