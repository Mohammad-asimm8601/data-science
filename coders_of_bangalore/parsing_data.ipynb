```python
# Parsing data

with open("initialdata.txt", encoding="utf-8") as f:
    data = f.read()
```

```python
data
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c) > 5]
```

```python
def parse_chunk(chunk):

        chunk = chunk.strip()
        sep_chunk = chunk.split('\n')
        username = sep_chunk[0]
        posts = int(sep_chunk[1].split(" post")[0].replace(",", ""))
        followers = int(sep_chunk[2].split(" follower")[0].replace(",", "").replace("K", "000").replace("M", "000000").replace(".", ""))
        following = int(sep_chunk[3].split(" following")[0].replace(",", ""))
        name = sep_chunk[4]
        if(len(sep_chunk) > 5):
            type_of_page = sep_chunk[5]
            bio = "\n".join(sep_chunk[6:])
        else:
            type_of_page = "Unknown"
            bio = ""
        # print(username, posts, followers, following, name, type_of_page, bio, sep="\n")
        return {"username":username, "posts":posts, "followers":followers, "following":following, "name":name, "type_of_page":type_of_page, "bio":bio, "type_of_page":type_of_page}
   
# print(parse_chunk(chunks[0]))
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
print(all_chunks)
```

```python
import json
s = json.dumps(all_chunks, indent = 4)
print(s)
```

 # Who has maximum post?

 ### Lets write code for maximum posts

```python
max = 0
chunk_with_max_post = None
for chunk in all_chunks:
    if(max < chunk["posts"]):
        max = chunk["posts"]
        chunk_with_max_post = chunk
print(chunk_with_max_post)
```
