```python
class Employee:
    company = "HP"
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    
    def printDetails(self):
        print(f"The company of {self.name} is {self.company}  with salary {self.salary}")
        
    @staticmethod
    def onTime():
        print("EveryThing is on time")

    @classmethod
    def printClassDetails(cls):
        print(f"The company is {cls.company}")

e1 = Employee("Asim", 47000)
e1.printDetails()
e1.onTime()
e1.printClassDetails()
```
